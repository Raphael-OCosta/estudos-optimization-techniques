### 1.1 **O primeiro problema.** A Boreal vende mesas a um preço `p`. A demanda é `q = 400 - 1,5·p` e o custo é R$ 80 por mesa. O lucro é `(p - 80)·q`. Calcule o lucro para `p` de 80 a 260, de 1 em 1, e ache o melhor preço **testando todos**. Desenhe a curva.

```python
import matplotlib.pyplot as plt
precos = np.arange(80, 261)
lucro = (precos - 80) * (400 - 1.5 * precos)
melhor = precos[np.argmax(lucro)]
plt.plot(precos, lucro); plt.axvline(melhor, ls="--", c="gray"); plt.xlabel("preço"); plt.ylabel("lucro"); plt.show()
print("melhor preço testando:", melhor, "| lucro:", lucro.max())
```

### 1.2 **O módulo básico.** Crie `basico.py` com:
- `busca_exaustiva(f, candidatos, maximizar=True)`: devolve `(melhor_x, melhor_valor)`;
- `otimo_quadratica(a, b)`: o vértice `-b/(2a)` de `a·x² + b·x + c`;
- `fatorial_2k(k)`: a matriz de planejamento com todas as combinações de -1 e +1 para k fatores (`itertools.product`), como DataFrame;
- `efeitos(plano, resposta)`: o efeito principal de cada fator, a média da resposta em +1 menos a média em -1.

_A célula grava `gabarito_otimizador/basico.py`: a 1ª linha (`%%writefile`) já vem pronta, digite o código embaixo dela._

```python
%%writefile ../gabarito_otimizador/basico.py
"""Ferramentas básicas: busca exaustiva, ótimo de quadrática e planejamento fatorial 2^k."""
from itertools import product

import numpy as np
import pandas as pd


def busca_exaustiva(f, candidatos, maximizar=True):
    valores = [f(x) for x in candidatos]
    i = int(np.argmax(valores) if maximizar else np.argmin(valores))
    return candidatos[i], valores[i]


def otimo_quadratica(a, b):
    return -b / (2 * a)


def fatorial_2k(k, nomes=None):
    nomes = nomes or [f"x{i + 1}" for i in range(k)]
    return pd.DataFrame(list(product([-1, 1], repeat=k)), columns=nomes)


def efeitos(plano, resposta):
    resposta = np.asarray(resposta, float)
    return {c: float(resposta[plano[c] == 1].mean() - resposta[plano[c] == -1].mean()) for c in plano.columns}
```

### 1.4 **Planejamento de experimentos (DOE).** A pintura das mesas tem 3 fatores: temperatura (baixa/alta), camadas (1/2) e tempo de secagem (curto/longo). Monte o fatorial 2³ e use a "fábrica" `defeitos = 20 - 3·temp - 5·camadas - 1·secagem + 2·temp·camadas + ruído` para gerar os resultados (`semente 1`). Calcule os efeitos principais e o da interação temperatura × camadas.

```python
r = np.random.default_rng(1)
plano = fatorial_2k(3, ["temperatura", "camadas", "secagem"])
y = 20 - 3 * plano.temperatura - 5 * plano.camadas - plano.secagem + 2 * plano.temperatura * plano.camadas + r.normal(0, .5, 8)
display(plano.assign(defeitos=y.round(2)))
print({k: round(v, 2) for k, v in efeitos(plano, y).items()})
inter = plano.temperatura * plano.camadas
print("interação temp×camadas:", round(y[inter == 1].mean() - y[inter == -1].mean(), 2))
```
